# Array Splitting in NumPy

Splitting is the process of breaking a single array into multiple sub-arrays along a specific axis. This is essentially the **reverse operation of concatenation**. NumPy provides three specialized functions for this:

---

### 1. Vertical Splitting: `np.vsplit`
* **Behavior**: Splits an array into multiple sub-arrays vertically (row-wise).
* **Axis**: It operates along **axis 0**.
* **Condition**: The total number of rows must be divisible by the number of splits.

### 2. Horizontal Splitting: `np.hsplit`
* **Behavior**: Splits an array into multiple sub-arrays horizontally (column-wise).
* **Axis**: It operates along **axis 1**.
* **Condition**: The total number of columns must be divisible by the number of splits.

### 3. Depth Splitting: `np.dsplit`
* **Behavior**: Splits an array along the 3rd dimension (depth).
* **Axis**: It operates along **axis 2**.
* **Requirement**: Only works on arrays with 3 or more dimensions.

```python
raw1=[10,20,30]
raw2=[40,50,60]
raw3=[70,80,90]

grid=np.array([raw1,raw2,raw3,raw1,raw2,raw1])

print(grid)
print("===========")
upper  , lower = np.vsplit(grid,[2])
print(upper)
print("===========")
print(lower)
```

```python
grid=np.array([[1,2,3,4,5],[5,6,8,7,8],[2,7,8,92,1],[5,6,6,7,8]])

print(grid)
left , right=np.hsplit(grid,[2])
print(left)
print("===========")
print(right)
```

# Array Stacking in NumPy

Stacking is the process of joining multiple arrays along a new or existing axis. It is the core method for combining data from different sources into a single structure. NumPy provides three specialized functions for this:

---

### 1. Vertical Stacking: `np.vstack`
* **Behavior**: Stacks arrays on top of each other (row-wise).
* **Axis**: It operates along **axis 0**.
* **Requirement**: Arrays must have the same number of columns.
* **Result**: Increases the number of rows.

### 2. Horizontal Stacking: `np.hstack`
* **Behavior**: Stacks arrays side-by-side (column-wise).
* **Axis**: It operates along **axis 1**.
* **Requirement**: Arrays must have the same number of rows.
* **Result**: Increases the number of columns.

### 3. Depth Stacking: `np.dstack`
* **Behavior**: Stacks arrays along the third dimension (depth).
* **Axis**: It operates along **axis 2**.
* **Common Use**: Combining separate color channels (Red, Green, Blue) to create a full RGB image.

```python
arr1=np.array([[1,2],[3,5]])
arr2=np.array([[7,2],[8,6]])
#print(arr1)
#print(arr2)
#np.concatenate
arr_col=np.concatenate((arr1,arr2),axis=0)
arr_col
```

# Transposing Arrays and Swapping Axe

Transposing is a special form of reshaping that similarly returns a view on the underlying data without copying anything. Arrays have the `transpose` method and the special `T` attribute:

```python
arr6=np.arange(15).reshape((3,5))
print(arr6)
print("=============")
print(arr6.T)
print("=============")
print(np.transpose(arr6))
print("=============")
print(arr6.swapaxes(0,1))
print("=============")
```
