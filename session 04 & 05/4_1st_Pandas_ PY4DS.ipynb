## **Creating Pandas Data Structures**

### Creating a Pandas Series

A Series is a one-dimensional array-like object containing a sequence of values (of similar types to NumPy types) of the same type and an associated array of data labels, called its *index*. The simplest Series is formed from only an array of data:

The string representation of a Series displayed interactively shows the index on the left and the values on the right. Since we did not specify an index for the data, a default one consisting of the integers `0` through `N - 1` (where `N` is the length of the data) is created. 

````python
obj = pd.Series([4, 7, -5, 3])
obj
````

get the array representation and index object of the Series via its `array` and `index` attributes, respectively:

````python
print(obj.array)
print("=" * 20)
print(obj.values)
print("=" * 20)
print(obj.index)
print("=" * 20)
print(obj.dtype)
print("=" * 20)
print(obj.ndim)
print("=" * 20)
print(obj.shape)
print("=" * 20)
print(obj.size)
````

# DataFrame

The Series and the DataFrame are intimately related. It's helpful to think of a DataFrame as actually being just a bunch of Series "glued together".

A DataFrame represents a rectangular table of data and contains an ordered, named collection of columns, each of which can be a different value type (numeric, string, Boolean, etc.). The DataFrame has both a row and column index; it can be thought of as a dictionary of Series all sharing the same index.

# Comparison: `del` vs. `.drop()` in Pandas

Both `del` and `.drop()` are used to remove data from a DataFrame, but they differ significantly in their functionality, flexibility, and impact on memory.
## 1. `del` (Python Keyword)

**Target**: Specifically designed to delete Columns only.

**Behavior**: It performs an In-place operation, meaning it modifies the original DataFrame immediately.

**Syntax:**
 ```python
 del df['column_name']
 ```

**Limitations**: 
* Cannot delete multiple columns at once.
* Cannot delete rows.
* Does not return any value (returns None).

## 2. `.drop()` (Pandas Method)

**Target**: Highly flexible; can delete both Rows and Columns.

**Behavior**: By default, it returns a new copy of the DataFrame and leaves the original untouched.
    
**Syntax**:
* **For Columns**:
```python
 df.drop(columns=['A', 'B'])
```
*OR*
```python
 df.drop(['A'], axis=1)
 ```

* **For Rows**:
```python
 df.drop(index=[0, 1])
 ```
*OR*
```python
df.drop([0], axis=0)
```
**In-place Modification**: Can modify the original DataFrame if you set the parameter inplace=True.

### Filling In Missing Data
Calling `fillna` with a constant replaces missing values with that value

````python
df=pd.DataFrame(np.random.randn(7, 3))
df.iloc[:4,1] = np.nan
df  
df.iloc[:2, 2] = np.nan
df.dropna()

df.fillna(0)
df.fillna(method='bfill')
````

Calling `fillna` with a dictionary, you can use a different fill value for each column:

````python
df.fillna({1: 0.5, 2: 0})
````

### Removing Duplicates


````python
data = pd.DataFrame({"k1": ["one", "two"] * 3 + ["two"],
                     "k2": [1, 1, 2, 3, 3, 4, 4]})
data
````

The DataFrame method `duplicated` returns a Boolean Series indicating whether each row is a duplicate (its column values are exactly equal to those in an earlier row) or not

````python
data.duplicated()
````

`drop_duplicates` returns a DataFrame with rows where the duplicated array is False filtered out

````python
data.drop_duplicates()
````

````python
data["v1"] = range(7)
data
data.drop_duplicates(subset=["k1"])
````

`duplicated` and `drop_duplicates` by default keep the first observed value combination. Passing `keep="last"` will return the last one:

````python
data.drop_duplicates(["k1", "k2"], keep="last")
````
