# Numpy Arrays
(NumPy stands for Numerical Python)


NumPy is a Python library used for working with arrays.It also has functions for working in domain of linear algebra and matrices.


**Why Use NumPy?**


In Python we have lists that serve the purpose of arrays, but they are slow to process.
**NumPy aims to provide an array object that is up to 50x faster than traditional Python lists.**
The array object in NumPy is called **ndarray**, it provides a lot of supporting functions that make working with ndarray very easy.

# Boolean Indexing

````python
names = np.array(["Bob", "Joe", "Will", "Bob", "Will", "Joe", "Joe"])
data = np.array([[4, 7], [0, 2], [-5, 6], [0, 0], [1, 2],[-12, -4], [3, 4]])

print(names)
print(data)
````

````python
print(names == "Bob")
print( data[names == "Bob"])
````

````python
print( data[names == "Bob", 1:])
print( data[names == "Bob", 1])
````

````python
print(names != "Bob")
````

## Boolean arithmetic operators like `&` (and) `|` (or)

````python
mask = (names == "Bob") | (names == "Will")
print(mask)
print(data[mask])
````

````python
data[data < 0] = 0
print(data)
````

````python

data[names != "Joe"] = 7
data
````

# Fancy Indexing
Fancy indexing is a term adopted by NumPy to describe indexing using integer arrays.

````python
arr = np.zeros((8, 4))
print(arr)
for i in range(8):
    arr[i] = i
print(arr)
````

## select a subset of the rows in a particular order

````python
print(arr[[4, 3, 0, 6]])
print(arr[[-3, -5, -7]])
````

````python
arr6 = np.arange(32)
print(arr6)
print(arr6.reshape(8, 4))
print(arr6.reshape(4, 8))
print(arr6.reshape(2, 16))
print(arr6.reshape(4, -1))  # Automatically determine the number of columns
print(arr6.reshape(-1, 4))  # Automatically determine the number of rows
````

````python
arr7 = arr6.reshape(8, 4)
print(arr7[[1, 5, 7, 2], [0, 3, 1, 2]])

print(arr7[[1, 5, 7, 2]][:, [0, 3, 1, 2]])
````

````python
print(arr7[[1, 5, 7, 2]][:, [0, 3, 1, 2]])

arr7[[1, 5, 7, 2], [0, 3, 1, 2]] = 0
print(arr7)
````

# Transposing Arrays and Swapping Axe

Transposing is a special form of reshaping that similarly returns a view on the underlying data without copying anything. Arrays have the `transpose` method and the special `T` attribute:

````python
arr8 = np.arange(15).reshape((3, 5))
print(f"arr8 = {arr8} \n")
print(arr8.T)
````

# matrix multiplication

````python
np.dot(arr8.T, arr8)
````

````python
arr8.T @ arr8
````

## swapping axes
le transposing with `.T` is a special case of swapping axes. ndarray has the method `swapaxes`, which takes a pair of axis numbers and switches the indicated axes to rearrange the data

````python
print(f"arr8 = \n {arr8} \n")
arr8.swapaxes(0, 1)
````

#  Pseudorandom Number Generation

The `numpy.random` module supplements the built-in Python `random` module with functions for efficiently generating whole arrays of sample values from many kinds of probability distributions. For example, you can get a 4 × 4 array of samples from the standard normal distribution using `numpy.random.standard_normal`

````python
samples = np.random.standard_normal(size=(4, 4))
print(samples)
````

````python
random_array = np.random.rand(3)
print(random_array)

random_array = np.random.random((3, 3))
print(random_array)

random_integers = np.random.randint(1, 10, (2, 2))
print(random_integers)
````

# File Input and Output with Arrays
NumPy is able to save and load data to and from disk in some text or binary formats.
```numpy.save``` and ```numpy.load``` are the two workhorse functions for efficiently <u>saving</u> and <u>loading</u> array data on disk. Arrays are saved by default in an uncompressed raw binary format with file <u>extension .npy</u>:

````python
arr = np.arange(10)
np.save("some_array", arr)
````

````python
np.load("some_array.npy")
````

You can save multiple arrays in an uncompressed archive using ```numpy.savez``` and passing the arrays as keyword arguments:

````python
np.savez("array_archive.npz", a=arr, b=arr)
````

````python
arch = np.load("array_archive.npz")
````
