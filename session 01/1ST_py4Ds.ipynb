## ndim 

```python
a= np.array(50)  # 0 dim
b= np.array([1,2,3,4,5]) #1-D
c= np.array([[0,1,3],[2,6,8]]) # 2-d
d= np.array([                      #3-D
            [
             [1,2,3],[5,6,7],[8,9,10]   
            ]
            ])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

```python
zeros=np.zeros((5,5),dtype=np.int64)
print(zeros)
```

```python
ones=np.ones((5,5))
print(ones)
print(ones.dtype)
```

```python
aranged_arr=np.arange(10)
rang_arr=[i for i in range(10)]
print(aranged_arr)
print(rang_arr)
```
